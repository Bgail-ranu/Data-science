```python
def transaction(transaction_list):
  """
  This function takes a list of transaction amounts and prints:
  - total sum of transactions
  - average amount
  - number of transactions
  """
  total_sum = sum(transaction_list)
  average_sum = total_sum / len(transaction_list)
  number_of_transactions = len(transaction_list)

  print(f"The total amount is = {total_sum}. The average amount is = {average_sum}. There are {number_of_transactions} transactions.")
```

```python
transaction_list = [500, 2000, 450, 660, 850]
transaction(transaction_list)
```

```python
transaction_list = input("Enter all your transactions: ").split()
transaction_list = [float(x) for x in transaction_list]
transaction(transaction_list)
```
