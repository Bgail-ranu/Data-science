```python
top_50 = df.sort_values('Total Amount', ascending =False).head(50)
top_50
```

```python
plt.figure(figsize =(6, 6))
plt.bar(x=df['Product Category'],
        height = df['Total Amount'],
        color = 'magenta',
        edgecolor = 'black',
        linewidth = 2
        )
plt.title('Top 50 Product Category by Total Amount', fontsize = 14, fontweight = 'bold')
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Total Amount', fontsize=12)
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize =(6,6))
sns.barplot(data = top_50,
            x = 'Product Category',
            y = 'Total Amount',
            palette = 'rocket'
            )
plt.title('Top 50 Product Category by Total Amount', fontsize = 14, fontweight = 'bold')
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Total Amount', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
top_gender = df.nlargest(10, 'Total Amount')[['Gender', 'Total Amount']]


plt.figure(figsize=(6, 6))

# Create bar plot - Seaborn style
sns.barplot(data= top_50,
            x='Gender',
            y='Total Amount',
            palette='rocket')  # Color palette

plt.title('Top 10 Gender by Total Amount (Seaborn)',
          fontsize=16,
          fontweight='bold')
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Total Amount', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
top_Products = df.nlargest(10, 'Total Amount')[['Product Category', 'Total Amount']]


plt.figure(figsize=(6, 6))

# Create bar plot - Seaborn style
sns.barplot(data= top_50,
            x='Product Category',
            y='Total Amount',
            palette='rocket')  # Color palette

plt.title('Top 10 Products by Total Amount (Seaborn)',
          fontsize=16,
          fontweight='bold')
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Total Amount', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,6))
sns.histplot(data=df,
             x='Product Category',
             bins = 20,
             kde = True,
             color = 'magenta',
             edgecolor = 'black',
             linewidth = 2
             )
plt.title('Product Category Distribution', fontsize = 16, fontweight = 'bold')
plt.ylabel('Frequency', fontsize = 12)
plt.xlabel('Product Category', fontsize = 12)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

# Boxplot comparing Total Amount across Product Category
sns.boxplot(data=df,
            x='Product Category',
            y='Total Amount',
            palette='rocket')

plt.title('Total Amount by Product Category',
          fontsize=16,
          fontweight='bold')
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Total Amount', fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

# Violin plot shows distribution shape + boxplot statistics
sns.violinplot(data=df,
               x='Product Category',
               y='Total Amount',
               palette='rocket')

plt.title('Total Amount Distribution by Product Category',
          fontsize=16,
          fontweight='bold')
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Total Amount', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

# Scatter plot with trend line
sns.regplot(data=df,
            x='Price per Unit',
            y='Total Amount',
            scatter_kws={'alpha':0.6, 's':100, 'edgecolor':'black'},
            line_kws={'color':'red', 'linewidth':2}
)

plt.title('Total Amount vs Price per Unit with Trend Line',
          fontsize=16,
          fontweight='bold')
plt.xlabel('Price per Unit', fontsize=12)
plt.ylabel('Total Amount', fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

correlation = df['Price per Unit'].corr(df['Total Amount'])
print(f"correlation coefficient: {correlation}")
```
