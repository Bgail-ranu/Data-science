```python
import pandas as pd
import numpy as np
import calendar
```

```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_csv('/content/drive/MyDrive/Warehouse_and_Retail_Sales.csv')
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df_c = df.drop(columns=['ITEM DESCRIPTION', 'ITEM CODE'])
#df_c['SUPPLIER'].fillna('Unknown', inplace=True)
df_c.fillna({'SUPPLIER': 'Unknown'}, inplace=True)
df_c = df_c.dropna(subset=['ITEM TYPE'])
df_c = df_c.dropna(subset=['RETAIL SALES'])
df.shape, df_c.shape, df_c.isnull().sum()
```

```python
df_c.duplicated().sum()
```

```python
df_c = df_c.drop_duplicates()
df_c.shape
```

```python
df['YEAR'].value_counts().sort_index()
```

```python
df.YEAR
```

```python
month_map = {
    1: "January", 2: "February", 3: "March",
    4: "April", 5: "May", 6: "June",
    7: "July", 8: "August", 9: "September",
    10: "October", 11: "November", 12: "December"
}

df_c['MONTH'] = df['MONTH'].map(month_map)

print(df_c[['MONTH']].tail())
```

```python
df_c.loc[182109:, 'MONTH'].head(10)
```

```python
df_c.columns
```

```python
df_c['ITEM TYPE'].value_counts().sort_index()
```

```python
item_analysis = df_c.groupby('ITEM TYPE')['RETAIL SALES'].agg([
    ('total_sales', 'sum'),
    ('Average_Sales', 'mean'),
    ('Minimum_Sales', 'min'),
    ('Maximum_Sales', 'max'),
    ('num_sales', 'count')
])
item_analysis = item_analysis.sort_values('total_sales', ascending=False)
print(item_analysis)
```

```python
df_c.to_csv('Warehouse_cleaned.csv', index=False)
```
